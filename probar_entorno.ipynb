# Participación de mujeres en el empleo registrado
Práctica con pandas. Fuente de datos: [(https://datos.gob.ar/dataset/porcentaje-de-mujeres-por-provincia-y-sector-de-actividad)].

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("datos/share_mujer_prod.csv")
df.head()
```

```python
df.info()
df.isna().sum()
```

```python
df["fecha"] = pd.to_datetime(df["fecha"])
df["clae2"] = df["clae2"].astype("category")
df["anio"] = df["fecha"].dt.year
```

## ¿Cómo cambió la participación de mujeres entre 2007 y 2023?

```python
por_anio = df.groupby("anio")["share_mujer"].mean()

por_anio.plot(marker="o", title="Participación de mujeres (promedio simple)")
plt.ylabel("Proporción")
plt.savefig("grafico_anual.png", bbox_inches="tight")
plt.show()
```

Entre 2007 y 2023 la proporción pasó de X a Y.
Limitación: el promedio es simple, sin ponderar por cantidad de trabajadores.
